```python
missing_percent = data.isnull().mean() * 100
cols_to_drop = missing_percent[missing_percent > 50].index
data = data.drop(columns=cols_to_drop)
```

```python
# Mean imputation (for normally distributed data)
data['likes'] = data['likes'].fillna(data['likes'].mean())

# Median imputation (for skewed data)
data['comments'] = data['comments'].fillna(data['comments'].median())

# Mean imputation again
data['watch_time_minutes'] = data['watch_time_minutes'].fillna(data['watch_time_minutes'].mean())
```

```python
# Detect numeric columns
num_cols = data.select_dtypes(include=['int64', 'float64']).columns

# Loop through numeric columns and impute
for col in num_cols:
    skewness = data[col].skew()
    if abs(skewness) < 0.5:
        data[col].fillna(data[col].mean(), inplace=True)
        print(f"{col}: Filled missing values with MEAN (normal distribution)")
    else:
        data[col].fillna(data[col].median(), inplace=True)
        print(f"{col}: Filled missing values with MEDIAN (skewed distribution)")

# Check if all missing values are filled
print("\n Missing values after imputation:\n")
print(data.isnull().sum())
```

```python
# Detect categorical columns
cat_cols = data.select_dtypes(include=['object']).columns

# Fill missing values with mode
for col in cat_cols:
    if data[col].isnull().sum() > 0:
        data[col].fillna(data[col].mode()[0], inplace=True)
        print(f"{col}: Filled missing values with MODE → '{data[col].mode()[0]}'")
    else:
        print(f"{col}: No missing values found.")

print("\n Missing values after categorical imputation:\n")
print(data.isnull().sum())
```

```python
# Forward Fill → fill missing values using the previous value
data_ffill = data.fillna(method='ffill')
print("Forward Fill applied - Missing values filled using previous row value.\n")
```

```python
# Backward Fill → fill missing values using the next value
data_bfill = data.fillna(method='bfill')
print("Backward Fill applied - Missing values filled using next row value.\n")
```

```python
print("Missing values after Forward Fill:\n", data_ffill.isnull().sum())
print("\nMissing values after Backward Fill:\n", data_bfill.isnull().sum())
```

```python
from sklearn.impute import KNNImputer

imputer = KNNImputer(n_neighbors=5)
data_knn = pd.DataFrame(imputer.fit_transform(data.select_dtypes(include=['int64', 'float64'])),
                        columns=data.select_dtypes(include=['int64', 'float64']).columns)

print("KNN Imputation applied — Missing values predicted using nearest neighbor values.\n")
print("Missing values after KNN Imputation:\n", data_knn.isnull().sum())
```
